## Part 3: How we use git in this course

In most projects, git is used two-way: `git pull` to get other people's changes, and `git commit` plus `git push` to send your own changes back to a shared repository, so everyone stays in sync. A typical loop looks like:
```powershell
git add some_file.py
git commit -m "Describe what changed"
git push
```

This course only uses the first half. Each week you'll `git pull` (as described in the README) to receive the new lab and you can start working on it right away. You can always go back to a previous lab with `git checkout`, but you won't be need to push anything back to this repository.

If you're already comfortable with git and want a personal record of your own progress, you're welcome to set up your own private repository and push your solved notebooks there (add a second remote, e.g. `git remote add origin <your-repo-url>`, then `git push origin labXX`).

## Part 6: Reading a traceback

You're going to see a lot of these this semester. A **traceback** is Python's report of exactly what went wrong and where. The skill isn't memorizing error types, it's reading the traceback calmly, bottom to top, to find the one line that actually matters.

Run the cell below. It's broken on purpose.

````python
def average_score(scores):
    total = sum(scores)
    return total / len(score)  # bug is here


class_scores = [88, 92, 79, 95, 84]
print(average_score(class_scores))
````

### Reading the traceback

Look at the **last line** first: that's the actual error (`NameError`, `TypeError`, etc.) and usually tells you what's wrong. Then look at the line just above it, prefixed with `--->` or similar: that's the line where it happened. Everything above that is the *call stack*: the chain of function calls that got you there, which matters more once your code is many functions deep, but for now the bottom two or three lines are what you need.

## Part 7: Cell execution order

Jupyter notebooks don't run top to bottom automatically. They run whatever cell *you* click "run" on, in whatever order you click. This is genuinely useful (you can re-run just one cell while experimenting), but it also means a notebook can look correct while secretly depending on an order that isn't written down anywhere.

Try this:

1. Run the first cell below (defines `mystery_value`).
2. Run the second cell below (uses `mystery_value`). It works.
3. Now go back and **change** the first cell so `mystery_value = 100` instead of `42`, but don't re-run it.
4. Re-run the second cell. What does it print?

````python
mystery_value = 42
````

````python
print(f"mystery_value is currently {mystery_value}")
````

If you changed the first cell but the second cell still printed `42`, that's the trap: the notebook's *displayed* code says `100`, but the running kernel still remembers the old value of `42` from before, because you never re-ran the first cell. Nothing about the notebook as it appears on screen tells you this happened.

This is why, from now on, **before you consider a notebook finished or submit it, use "Restart Kernel and Run All"** (in the Kernel menu). It wipes the kernel's memory and re-runs every cell top to bottom, exactly once, in order. This is the only way to be sure the notebook actually works the way it looks like it does, and not just because of leftover state from an order you've since forgotten.

Try it now: restart the kernel and run all cells in this notebook from the top, and confirm `mystery_value` prints `100` (matching what the first cell now says).

## Part 8: Running a plain Python script

Everything so far has run inside notebook cells. But `minitorch/` files themselves are never run as notebooks — they're plain `.py` files that get *imported*. It's worth seeing that distinction directly, since it's how you'll run things outside of Jupyter all semester (and how you'll eventually run full training scripts, not just notebook experiments).

### Steps

1. In your editor (VS Code or otherwise), create a new file at the root of your repo called `hello.py` (not inside `minitorch/` or `labs/`) with this content:
   ```python
   def greet(name):
       return f"Hello, {name}!"


   if __name__ == "__main__":
       print(greet("world"))
   ```
2. In your **terminal**, from the repo's root folder, with `dl-labs` activated, run it as a script:
   ```powershell
   python hello.py
   ```
   You should see `Hello, world!` printed.
3. Now try importing it instead of running it, from a Python session. In your terminal:
   ```powershell
   python
   ```
   This opens an interactive Python prompt. Then:
   ```python
   from hello import greet
   greet("class")
   ```
   Type `exit()` to leave the interactive prompt when you're done.

Notice `if __name__ == "__main__":` only ran when you executed the file directly with `python hello.py`, not when you imported it. That's the same mechanism that lets `minitorch/utils.py` be safely imported (as you did in Part 4) without anything inside it running on its own. You can delete `hello.py` once you're done — it was just for practice.
